### I-1

```python
# I-1
import math

class Cylindricity:
    r = 0
    h = 0
    v = 0
    def __init__(self, r, h):
        self.r = r
        self.h = h
        self.v = math.pi * r ** 2 * h

input_r, input_h= map(float, input("Enter radius and height: ").split())
cyl = Cylindricity(input_r, input_h)
print(f"{cyl.v}")
```

```python
# I-2
distance = lambda x1, y1, x2, y2: ((x2 - x1) ** 2 + (y2 - y1) ** 2) ** 0.5
input_data = input("input x1 y1 x2 y2: ")
x1, y1, x2, y2 = map(float, input_data.split())
print(f"distance: {distance(x1, y1, x2, y2)}")
```

```python
# I-3
def each_number_sum(n):
    sum = 0
    for i in n:
        sum += int(i)
    return sum

input_number = input("Enter a number: ")
print(each_number_sum(input_number))
```

```python
# I-6
def is_lunar_year(year):
    if year % 4 == 0 and year % 100 != 0 or year % 400 == 0:
        return True
    else:
        return False
input_year = int(input("Enter a year: "))
if is_lunar_year(input_year):
    print("是")
else:
    print("否")
```

### I-7

```python
import math

def can_split_to_squares(n, m, k):
    if (k < 0 or k > n * m):
        return False
    for i in range(math.ceil(n / 2)):
        if(i * m == k or (n - i) * m == k):
            return True
    for j in range(math.ceil(m / 2)):
        if(n * j == k or n * (m - j) == k):
            return True
    return False

n, m, k = map(int, input().split())
if (can_split_to_squares(n, m, k)):
    print("是")
else:
    print("否")
```

### I-8

```python
def square_sum(n, list):
    sum = 0
    for i in range(n):
        sum += list[i] ** 2
    return sum

input_n = int(input())
input_list = list(map(int, input().split()))
print(square_sum(input_n, input_list))
```
